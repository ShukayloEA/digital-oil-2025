```python
def date_to_timestamp(x):
    return datetime.datetime.strptime(x, "%Y-%m-%d %H:%M:%S").timestamp()
```

```python
def dataframe_date_to_timestamp(x_list):
    y_list = []
    for x in x_list:
        y_list.append(datetime.datetime.strptime(x, "%Y-%m-%d %H:%M:%S").timestamp())
    return y_list
```
